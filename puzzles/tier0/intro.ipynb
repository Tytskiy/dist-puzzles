```python
import flock
```

## 1. Ping

2 workers. Rank 0 sends `"ping"` to rank 1. Rank 1 receives it and returns the message. Rank 0 returns `None`.

Expected: `[None, "ping"]`

```python
@flock.distribute(workers=2, seed=0)
async def ping():
    if flock.get_rank() == 0:
        await flock.send(1, "ping")
        return None
    return await flock.recv(0)


ping()
```

## 2. Sum

4 workers. Each rank contributes its own `rank` to an `all_reduce` with `"sum"`. Return the result.

Expected: `[6, 6, 6, 6]`

```python
@flock.distribute(workers=4)
async def sum_ranks():
    return await flock.all_reduce(flock.get_rank(), "sum").wait()


sum_ranks()
```

## 3. Gather

4 workers. Each rank contributes the list `[1, 2, 3]` to `all_gather`. Return the gathered result.

Expected: `[[1, 2, 3], [1, 2, 3], [1, 2, 3], [1, 2, 3]]` on every rank.

```python
@flock.distribute(workers=4)
async def gather_lists():
    return await flock.all_gather([1, 2, 3]).wait()


result = gather_lists()

[id(r) for r in result]
```
